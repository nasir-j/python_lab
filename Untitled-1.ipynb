```python
import numpy as np
arr = np.random.randint(10,101, size = (20,5))
print(arr)

#mean
mean = np.mean(arr, axis= 0)
#std dev
std = np.std(arr, axis= 0)

print(mean)
print(std)


row_sum = np.sum (arr, axis = 1)
max_row = np.argmax(row_sum)

print([max_row])

print("Row number:", max_row + 1)
print("Highest sum:", row_sum[max_row])
```

```python
# Simulate scores of 50 students in 3 subjects, values between 0 and 100.
import numpy as np
arr = np.random.randint(0,101, size=(50,3))
print("Score:\n", arr)

#avg of each student

avg = np.mean(arr, axis=1)
print(avg)

#top 5 students

top5 = np.argsort(avg)[-5:][::-1]
print ("top5", top5)
for i in top5:
    print("student", i+1, "avg", avg[i])
```

```python
import numpy as np
av = np.array([10,12,30,50,60])
pv = np.array([12,10,28,53,56])
mse = np.mean((av-pv)**2)
rmse = np.sqrt(mse)

print("MSE:", mse)
print("RMSE:", rmse)
```

```python
import numpy as np
basket = np.random.randint(0,2, size=(200, 10))
print("Basket:", basket)

count = np.sum(basket, axis = 0)
print(count)

co_occurence = basket.T @ basket
print(co_occurence)
```

```python
import numpy as np

# Generate 100 samples with 2 features
X = np.random.normal(100, 2, (100, 2))

# Generate target values
y = 3 * X[:, 0] + 5 * X[:, 1] + 4 + np.random.randn(100) * 0.2

# Add bias/intercept column
X_bias = np.column_stack((np.ones(100), X))

# Closed-form solution
beta = np.linalg.inv(X_bias.T @ X_bias) @ X_bias.T @ y

print("Coefficients:")
print(beta)
```
